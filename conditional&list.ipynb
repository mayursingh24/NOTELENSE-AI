```python
marks = float(input("Enter marks: "))
if marks >= 80:
    print("Grade: A")
elif marks >= 60 and marks < 80:
    print("Grade: B")
elif marks >= 50 and marks < 60:
    print("Grade: C")
```

```python
name= input("eneter your name:")
course= input("enter course:")
roll=input("enter roll:")
branch=input("enter branch:")
s1=int(input("enter marks for sub 1:"))
s2=int(input("enter marks for sub 2:"))
s3=int(input("enter marks for sub 3:"))
s4=int(input("enter marks for sub 4:"))
print("Hello:-",name)
print("Course- ",course)
print("roll- ",roll)
print("branch- ",branch)
print("your % -", (s1+s2+s3+s4)/4)
l=[s1,s2,s3,s4]
n=0
for i in l :
  n+=1
  p=i
  print("For sub ",n)
  if p >= 80:
    print("Grade-A")
  elif p>=60:
    print("Grade-B")
  elif p>=40:
      print("Grade-C")
  elif p>=30:
    print("Grade-D")
  elif p>=23:
    print("Grade-E")
  else:
    print("Grade-F")
```

```python
l =[1,2,3,4,4, "sudh", "kumar", 324,34.56, "abc"]
```

```python
num = []
str =[]
flt = []
for i in l :
    if type(i) == int :
        num.append(i)
    elif type(i) == float :
        flt.append(i)
    else :
        str.append(i)
print(num)
print(str)
print(flt)
```

```python
num = input("Enter a number")
if num == num[::-1]:
    print("Yes its a palindrome")
else:
    print("No, its not a palindrome")
```
